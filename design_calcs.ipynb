## Shell-and Tube Heat Exchanger Design Calculations
Water-to-water, counter flow configuration

```python
# Design inputs
m_hot = 0.5    # kg/s, hot water mass flow rate
m_cold = 0.75  # kg/s, cold water mass flow rate
T_hot_in = 80.0  # °C
T_hot_out = 50.0  # °C
T_cold_in = 20.0  # °C
T_cold_out = 40.0  # °C
cp = 4186 # J/(kg·K), specific heat capacity of water
```

## Heat Duty Q
Using the hot side energy balance: Q = m_hot * cp *(T_hot_in - T_hot_out)

```python
Q = m_hot * cp * (T_hot_in - T_hot_out)
print(f"Heat duty Q = {Q/1000:.1f} kW")
```

## Log Mean Temperature Difference (LMTD)
Counter flow arrangement - hot and cold streams flow in opposite directions.
delta_T1 = T_hot_in - T_cold_out
delta_T2 = T_hot_out - T_cold_in
LMTD = (delta_T1 - delta_T2) / ln(delta_T1/delta_T2)

```python
import math
delta_T1 = T_hot_in - T_cold_out
delta_T2 = T_hot_out - T_cold_in

LMTD = (delta_T1 - delta_T2) / math.log(delta_T1 / delta_T2)
print(f"delta_T1 = {delta_T1} °C,")
print(f"delta_T2 = {delta_T2} °C,")
print(f"LMTD = {LMTD:.1f} °C")
```

## Overall Heat Transfer Coefficient and Required Area
U is assumed based on typical values for water to water shell and tube exchangers.
Required area: A = Q / (U * LMTD)


```python
U = 1000 # W/(m²·K), assumed for water-to-water service
A_required = Q / (U * LMTD)
print(f"U = {U} W/(m²·K)")
print(f"Required area A = {A_required:.2f} m²")
```

## Tube Sizing
Given tube OD, wall thickness and length, calculate area per tube and required tube count.

```python
tube_OD = 0.01905   # m, 19.05 mm
tube_length = 0.5   # m
num_tubes = 60

area_per_tube = math.pi * tube_OD * tube_length
total_tube_area = area_per_tube * num_tubes

print(f"Area per tube = {area_per_tube:.4f} m²")
print(f"Total tube area ({num_tubes} tubes) = {total_tube_area:.2f} m²")
```

## Shell Sizing
Tube layout on square pitch, bundle diameter estimated from tube count and pitch, shell ID sized with clearance over the bundle.

```python
tube_pitch = 0.0238  # m, 23.8 mm square pitch
bundle_clearance = 0.038  # m, 38 mm clearance between tube bundle and shell

# Simplified bundle diameter estimate for a square pitch layout
bundle_diameter = tube_pitch * math.sqrt(num_tubes) * 1.1 # rough packing factor
Shell_ID = bundle_diameter + bundle_clearance

print(f"Estimated bundle diameter = {bundle_diameter*1000:.0f} mm")
print(f"Shell ID = {Shell_ID*1000:.0f} mm")
```

Simplified estimate (203mm bundle/241mm shell) closely matches the detailed tube layout values used in the actual CAD model (207mm bunde/245mm shell ID).
